```python
eksik_veriler = df[df.isnull().any(axis=1)]
#isnull ile kontrol sağladık axis = 1 ile en az bir tane boş aradık
eksik_veriler.head(20)
```

```python
print(f"Doldurmadan önce : \n {df[["indirim_orani","musteri_puani"]].isnull().sum()}")
#isnull sum ile kontrol sağlıyoruz
```

```python
df["indirim_orani"] = df["indirim_orani"].fillna(df["indirim_orani"].median())
#fillna fonksiyonu ile seçtiğimiz sütundaki boş satırları o sütunun medyanı ile doldurduk
df["musteri_puani"] = df["musteri_puani"].fillna(df["musteri_puani"].median())
```

```python
# Fillna Fonksiyonu sayesinde mod  ile doldurma
df["odeme_turu"] = df["odeme_turu"].fillna(df["odeme_turu"].mode()[0])
df["musteri_tipi"] = df["musteri_tipi"].fillna(df["musteri_tipi"].mode()[0])
```

```python
sehir_duzeltme_haritasi = {
    "Istanbul": "İstanbul",
    "Izmir": "İzmir"
}
df["sehir"] = df["sehir"].replace(sehir_duzeltme_haritasi)
df["sehir"].unique()
```

```python
# Temizlikten sonra benzersiz şehir sayısı
print("Temizlikten sonra:", df["sehir"].nunique())
```

```python
df["kategori"].unique()
```

```python
df["kategori"] = df["kategori"].str.strip()      # Baştaki ve sondaki boşlukları sil
df["kategori"] = df["kategori"].str.title()      # İlk harfi büyük yap
```

```python
kategori_duzeltme_haritasi = {
    "Ev&Yaşam": "Ev Yaşam",
    "Ev Ve Yaşam": "Ev Yaşam",
    "Kozmetik ": "Kozmetik"
}
df["kategori"] = df["kategori"].replace(kategori_duzeltme_haritasi)
```

```python
df["kategori"].value_counts()
```

```python
hatali_kayitlar = df[
    (df["birim_fiyat"] <= 0) |
    (df["toplam_tutar"] <= 0) |
    (df["teslimat_gunu"] < 0) |
    (df["musteri_puani"] > 5)
]
#hatalı kayıtları filtreleme yaptık
```

```python
print("Hatalı kayıt sayısı:", hatali_kayitlar.shape[0])
```

```python
df = df[
    (df["birim_fiyat"] > 0) &
    (df["toplam_tutar"] > 0) &
    (df["teslimat_gunu"] >= 0) &
    (df["musteri_puani"] <= 5)
]
#veri setinden çıkartıyoruz
```

```python
print("Temizlendikten sonraki kayıt sayısı:", df.shape[0])
```

```python
# Q1 ve Q3 hesapla
Q1 = df["birim_fiyat"].quantile(0.25)
Q3 = df["birim_fiyat"].quantile(0.75)
```

```python
# IQR hesapla
IQR = Q3 - Q1
```

```python
# Alt ve üst sınır
alt_sinir = Q1 - 1.5 * IQR
ust_sinir = Q3 + 1.5 * IQR
```

```python
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Alt sınır:", alt_sinir)
print("Üst sınır:", ust_sinir)
```

```python
aykiri_degerler = df[
    (df["birim_fiyat"] < alt_sinir) |
    (df["birim_fiyat"] > ust_sinir)
]
```

```python
print("Aykırı değer sayısı:", aykiri_degerler.shape[0])
```
