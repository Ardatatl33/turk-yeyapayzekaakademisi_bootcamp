## kayıp veri (missing value)

```python
import pandas as pd
df = pd.read_csv("/content/e_ticaret_veri_seti.csv")
```

```python
#eksik verileri  tespit etme
df.isnull().sum()
```

```python
# eksik veri oranlarını hesapla
eksik_sayilari = df.isnull().sum()
eksik_oranlari = 100 * (eksik_sayilari / len(df))
eksik_oranlari
```

```python
#eksik veri içeren satırları incele
eksik_veriler = df[df.isnull().any(axis = 1)]
eksik_veriler.head(10)
```

```python
#eksik veri silme
df_silinen = df.dropna()
```

```python
df_silinen.shape
```

```python
#6. sayısal eksik verilerin ortalamayla doldurulması
df_ortalama = df.copy()
sayisal_sutunlar = df_ortalama.select_dtypes(include=["int64","float64"]).columns
sayisal_sutunlar
```

```python
for sütun in sayisal_sutunlar:
 df_ortalama[sütun] = df_ortalama[sütun].fillna(df_ortalama[sütun].mean())
```

```python
df_ortalama[sayisal_sutunlar].isnull().sum()
```

```python
# sayısal eksik verilerin medyan doldurulması
df_medyan = df.copy()
for sütun in sayisal_sutunlar:
  df_medyan[sütun] = df_medyan[sütun].fillna(df_medyan[sütun].median())
df_medyan[sayisal_sutunlar].isnull().sum()
```

```python
# sayısal eksik verilerin mod doldurulması
df_mod = df.copy()
for sütun in sayisal_sutunlar:
  mod_degeri = df_mod[sütun].mode()[0]
  df_mod[sütun] = df_medyan[sütun].fillna(mod_degeri)
df_mod[sayisal_sutunlar].isnull().sum()
```

```python
# kategorik verileri mod ile doldurma
df_kategorik = df.copy()
kategorik_sutunlar = df_kategorik.select_dtypes(include=["object"]).columns
kategorik_sutunlar
```

```python
for sütun in kategorik_sutunlar:
  mod_degeri = df_kategorik[sütun].mode()[0]
  df_kategorik[sütun] = df_kategorik[sütun].fillna(mod_degeri)
df_kategorik[kategorik_sutunlar].isnull().sum()
```

```python
# kategorik eksik verileri sabitle doldurma
df_sabit_kategorik = df.copy()
for sütun in kategorik_sutunlar:
  df_sabit_kategorik[sütun] = df_sabit_kategorik[sütun].fillna("Belirtilmemiş")
df_sabit_kategorik[kategorik_sutunlar].isnull().sum()
```

```python
from numpy import median
df_guncel = df.copy()
for sutun in sayisal_sutunlar:
  df_guncel[sutun] = df_guncel[sutun].fillna(df[sutun].median())
```

```python
for sutun in kategorik_sutunlar:
  mod_degeri = df_guncel[sutun].mode()[0]
  df_guncel[sutun] = df_guncel[sutun].fillna(mod_degeri)
```

```python
df_guncel.isnull().sum()
```

```python
# kaydetme
kayıt_veri_seti = "eksik_veri_seti_düzenlendi.csv"
df_guncel.to_csv(kayıt_veri_seti, index=False)
```

## Tekrarlayan ve Tutarsız kayıtlar
Amaç :
  1. Veri setindeki tekrarlayan ve tutarsız verileri tepit edip temizlemek
  2. Eğitimde kullanılan güncel verileri kaydetmek

Adımlar:
  1. Veri setini yükle
  2. Tekrarlayan kayıtları tespit etmek
  3. Tekrarlayan kayıtları incelemek
  4. Tekrarlayan kayıtları temizlemek
  5. Tutarsız kategorik değerleri incelemek
  6. Şehir sutunundaki tutarsız değerleri düzeltmek
  7. Kategori sutunundaki tutarsız değerleri düzeltmek
  8. Ödeme Türü sutunundaki tutarsız değerleri düzeltmek
  9. Müşteri Tipi sutunundaki tutarsız değerleri düzeltmek
  10. Mantıksal olarak hatalı kayıtları incele
  11. Mantıksal olarak hatalı kayıtları temizleme
  12. Temizlenmiş veri setini kaydetme

```python
import pandas as pd
```

```python
df = pd.read_csv("/content/e_ticaret_veri_seti_veri_tipleri_duzenlendi.csv")
```

```python
# Tekrarlayan kayıtları tespit etmek
tekrarlayan_kayitlar = df[df.duplicated()]
len(tekrarlayan_kayitlar)
```

```python
# Tekrarlayan kayıtları incelemek
tekrarlayan_kayitlar.head(10)
```

```python
# 4.tekrarlayan kayıtları temizlemek
df = df.drop_duplicates()
df.shape
```

```python
# 5. Tutarsız kategorik değerleri incelemek
df["sehir"].unique()
```

```python
df["kategori"].unique()
```

```python
df["odeme_turu"].unique()
```

```python
#yardımcı fonksiyon : metinleri normalize edelim
#büyük küçük harf dönüşümleri, boşlukların kaldırılması, Türkçe karakter kaynaklı farkların düzeltilmesi
def metin_normalize_et(series):

  return(
      series.astype(str)
      .str.strip()
      .str.lower()
      .str.replace("i̇", "i", regex=False)
      .str.replace("ı", "i", regex=False)
      .str.replace("ö", "o", regex=False)
      .str.replace("ü", "u", regex=False)
      .str.replace("ç", "c", regex=False)
      .str.replace("ş", "s", regex=False)
      .str.replace("ğ", "g", regex=False)
  )
```

```python
#6. Şehir sütunundaki tutrasız verileri düzeltme
df["sehir"] = metin_normalize_et(df["sehir"])
df["sehir"].unique()
```

```python
sehir_duzeltme_haritasi = {
    "istanbul": "İstanbul",
    "ankara": "Ankara",
    "izmir": "İzmir",
    "antalya": "Antalya",
    "konya": "Konya",
    "bursa": "Bursa",
    "adana": "Adana",
    "gaziantep": "Gaziantep",
    "mersin": "Mersin",
    "kayseri": "Kayseri",
    "eskisehir": "Eskişehir",
    "samsun": "Samsun",
    "trabzon": "Trabzon",
    "kocaeli": "Kocaeli"
}
df["sehir"] = df["sehir"].replace(sehir_duzeltme_haritasi)
df["sehir"].unique()
```

```python
df["kategori"] = metin_normalize_et(df["kategori"])
df["kategori"].unique()
```

```python
kategori_duzeltme_haritasi = {
    "elektronik": "Elektronik",
    "kitap": "Kitap",
    "kozmetik": "Kozmetik",
    "kozmetık": "Kozmetik",
    "spor": "Spor",
    "ev yasam": "Ev Yaşam",
    "ev ve yasam": "Ev Yaşam",
    "ev&yasam": "Ev Yaşam",
    "ofis": "Ofis"
}
df["kategori"] = df["kategori"].replace(kategori_duzeltme_haritasi)
df["kategori"].unique()
```

```python
df["odeme_turu"] = metin_normalize_et(df["odeme_turu"])
df["odeme_turu"].unique()
```

```python
odeme_turu_duzeltme_haritasi = {
    "kredi karti": "Kredi Kartı",
    "kart": "Kredi Kartı",
    "banka karti": "Banka Kartı",
    "dijital cuzdan": "Dijital Cüzdan",
    "kapida odeme": "Kapıda Ödeme",
    "havale": "Havale/EFT",
    "eft": "Havale/EFT",
    "havale/eft": "Havale/EFT"
}
df["odeme_turu"] = df["odeme_turu"].replace(odeme_turu_duzeltme_haritasi)
df["odeme_turu"].unique()
```

```python
df["musteri_tipi"] = metin_normalize_et(df["musteri_tipi"])
df["musteri_tipi"].unique()
```

```python
musteri_tipi_duzeltme_haritasi = {
    "mevcut": "Mevcut",
    "yeni": "Yeni",
    "vip": "VIP"
}
df["musteri_tipi"] = df["musteri_tipi"].replace(musteri_tipi_duzeltme_haritasi)
df["musteri_tipi"].unique()
```

```python
df["musteri_tipi"].value_counts()
```

```python
# mantıksal olarak hatalı kayıtları incele
#negatif yada anlamsız değerler bzeı kayıtlarda veri kalitesini düşürür
hatali_kayitlar = df[
    (df["adet"] < 0) |
    (df["birim_fiyat"] <= 0) |
    (df["toplam_tutar"] <= 0) |
    (df["teslimat_gunu"] < 0) |
    (df["musteri_puani"] < 0) |
    (df["musteri_puani"] > 5)
]
len(hatali_kayitlar)
```

```python
pd.set_option("display.max_columns", None)
hatali_kayitlar
```

```python
# mantıksal hataları temizle
df = df[
    (df["adet"] > 0) &
    (df["birim_fiyat"] > 0) &
    (df["toplam_tutar"] > 0) &
    (df["teslimat_gunu"] >= 0) &
    (df["musteri_puani"] >= 0) &
    (df["musteri_puani"] <= 5)
]
df.shape
```

```python
df.to_csv("e_ticaret_veri_seti_tekrarlayan_kayıtlar_düzenlendi.csv")
```
